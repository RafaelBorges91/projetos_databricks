```sql
SELECT* FROM `workspace`.`default`.`series_tv` limit 100;


-- Os primeiros dez registros cadastrados
SELECT id, Titulo, Ano, `Indicação`, IMdb
FROM series_tv
ORDER BY id
LIMIT 10;


------------------------------------------------
---- Soma das Séries e o intervalo de lancamento: 

SELECT COUNT(*) AS total_series,
       MIN(Ano) AS ano_mais_antigo,
       MAX(Ano) AS ano_mais_recente
FROM series_tv;

-----------------------------------------------
--- Filtrando títulos que possuem a palavra "Breaking":
SELECT Id, Titulo, Ano
FROM series_tv
WHERE Titulo LIKE '%Breaking%'
ORDER BY titulo;

-----------------------------------------------
----- Filtro: series lancadas entre 2015 e 2021

SELECT Titulo, Ano, IMDb
FROM series_tv
WHERE Ano BETWEEN 2015 AND 2021
ORDER BY Ano DESC, Titulo
LIMIT 20;

----------------------------------------------------
----- Filtro: series lancadas entre 2000 e 2015

SELECT Titulo, Ano, IMDb
FROM series_tv
WHERE Ano BETWEEN 2000 AND 2015
ORDER BY Ano DESC, Titulo
LIMIT 20;

---------------------------------------------------
----- Filtro por títulos que não possuem nulos, 
----- ordenando pela IMDb, descrente: 

SELECT Titulo, Ano, IMDb
FROM series_tv
WHERE IMDb IS NOT NULL
ORDER BY IMDb DESC, titulo, id
LIMIT 10;

-----------------------------------------------------
----- Filtro: series lancadas entre 2000 e 2015

SELECT DISTINCT `Indicação`
FROM series_tv
ORDER BY `Indicação`;

-----------------------------------------------------
------- QTD de series em cada plataforma: 

SELECT SUM(Netflix) AS Netflix,
       SUM(Hulu) AS Hulu,
       SUM(`Prime Video`) AS Prime_Video,
       SUM(`Disney+`) AS `Disney+`
FROM series_tv;

------------------------------------------------------
------- Distribuicao de registros por faixa etaria: 

SELECT CASE WHEN `Indicação` IS NULL THEN 'Nao informada'
            ELSE `Indicação` END AS faixa_etaria,
       COUNT(*) AS total_series
FROM series_tv
GROUP BY `Indicação`
ORDER BY total_series DESC;

---------------------------------------------------------
------- QTD de series lancadas em cada decada:  
SELECT FLOOR(Ano / 10) * 10 AS decada,
       COUNT(*) AS total_series
FROM series_tv
GROUP BY FLOOR(Ano / 10) * 10
ORDER BY decada;

----------------------------------------------------------
--- Quais anos concentram o maior numero de registros:
SELECT Ano, COUNT(*) AS total_series
FROM series_tv
GROUP BY Ano
ORDER BY total_series DESC, Ano DESC
LIMIT 10;


-----------------------------------------------------------
----- Identificando a media do IMDb em cada plataforma: 
SELECT 'Netflix' AS plataforma, COUNT(*) AS Titulos,
       COUNT(IMDb) AS titulos_avaliados,
       ROUND(AVG(CAST(SPLIT_PART(IMDb, '/', 1) AS DOUBLE)), 2) AS media_imdb
FROM series_tv WHERE Netflix = 1
UNION ALL
SELECT 'Hulu', COUNT(*), COUNT(IMDb), ROUND(AVG(CAST(SPLIT_PART(IMDb, '/', 1) AS DOUBLE)), 2)
FROM series_tv WHERE hulu = 1
UNION ALL
SELECT 'Prime Video', COUNT(*), COUNT(IMDb), ROUND(AVG(CAST(SPLIT_PART(IMDb, '/', 1) AS DOUBLE)), 2)
FROM series_tv WHERE `Prime Video` = 1
UNION ALL
SELECT 'Disney+', COUNT(*), COUNT(IMDb), ROUND(AVG(CAST(SPLIT_PART(IMDb, '/', 1) AS DOUBLE)), 2)
FROM series_tv WHERE `Disney+` = 1;



```
